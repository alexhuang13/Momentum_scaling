# Blowup under increasing momentum

This diagnostic uses the schedule $\rho_k = 1 - 1/(k+1)$ and compares the median population loss of SGD, Polyak momentum, and Nesterov momentum. It is intentionally separate from the fixed-$\rho$ stability-boundary experiments.

Set `PAPER_RUN=True` for the saved paper configuration. The quick configuration is intended only to verify the environment and execution path.

```python
from pathlib import Path

from momentum_scaling.blowup import BlowupRuntime, blowup_experiment
from momentum_scaling.core import resolve_device, resolve_dtype, seed_everything
from momentum_scaling.plotting import plot_blowup

PAPER_RUN = False
SEED = 42
OUTPUT_DIR = Path("results/blowup/increasing_momentum")

DEVICE = resolve_device("auto")
DTYPE = resolve_dtype("auto", DEVICE)
seed_everything(SEED)
print(f"device={DEVICE}, dtype={DTYPE}, output={OUTPUT_DIR.resolve()}")
```

## Run the diagnostic

The numerical implementation lives in `src/momentum_scaling/blowup.py`; this notebook only selects the preset, runs it, and displays the result.

```python
runtime = BlowupRuntime(
    device=DEVICE,
    dtype=DTYPE,
    seed=SEED,
    output_dir=OUTPUT_DIR,
)
payload = blowup_experiment(runtime, quick=not PAPER_RUN)
payload["results"]
```

```python
plot_blowup(payload, OUTPUT_DIR)
```
